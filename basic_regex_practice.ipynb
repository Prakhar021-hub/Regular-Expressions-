# 📘 Regex Metacharacters Cheat Sheet

This is a quick reference for **regular expression (regex) metacharacters**.

---

## 🔹 Matching Basics
| Metachar | Meaning / Use | Example |
|----------|---------------|---------|
| `.` | Matches **any single character** except newline `\n`. | `a.c` → `abc`, `axc` |
| `^` | Matches the **start** of a string. | `^Hello` → `"Hello world"` |
| `$` | Matches the **end** of a string. | `world$` → `"Hello world"` |
| `[]` | Matches **any one character** inside. | `[aeiou]` → vowel |
| `[^]` | Matches any character **NOT** inside. | `[^0-9]` → non-digit |
| `-` | Defines a **range** in brackets. | `[a-z]` → lowercase letters |
| `|` | Acts like **OR**. | `cat|dog` → matches `cat` or `dog` |

---

## 🔹 Grouping
| Metachar | Meaning / Use | Example |
|----------|---------------|---------|
| `()` | Groups expressions / captures them. | `(ab)+` → `ab`, `abab` |
| `(?: )` | Groups without capturing. | `(?:ab)+` |

---

## 🔹 Quantifiers
| Metachar | Meaning / Use | Example |
|----------|---------------|---------|
| `?` | **0 or 1 occurrence** (optional). | `colou?r` → `color`, `colour` |
| `*` | **0 or more occurrences**. | `go*` → `g`, `go`, `goo` |
| `+` | **1 or more occurrences**. | `go+` → `go`, `goo` |
| `{n}` | Exactly **n times**. | `a{3}` → `aaa` |
| `{n,}` | **n or more times**. | `a{2,}` → `aa`, `aaa` |
| `{n,m}` | Between **n and m times**. | `a{2,4}` → `aa`, `aaa`, `aaaa` |

---

## 🔹 Escapes & Character Classes
| Metachar | Meaning / Use | Example |
|----------|---------------|---------|
| `\` | Escape special char. | `\.` → matches `.` |
| `\d` | Digit (0–9). | `\d{3}` → `123` |
| `\D` | Non-digit. | `\D+` → `abc` |
| `\w` | Word char (letters, digits, `_`). | `\w+` → `hello_123` |
| `\W` | Non-word char. | `\W` → `@` |
| `\s` | Whitespace (space, tab, newline). | `\s+` → `" "` |
| `\S` | Non-whitespace. | `\S+` → `word` |

---

## 🔹 Boundaries
| Metachar | Meaning / Use | Example |
|----------|---------------|---------|
| `\b` | Word boundary. | `\bcat\b` → `cat` only |
| `\B` | Not a word boundary. | `\Bcat\B` → in `educational` |

---

## 🔹 Lookarounds
| Metachar | Meaning / Use | Example |
|----------|---------------|---------|
| `(?=...)` | Positive lookahead. | `\d(?=px)` → digit before `px` |
| `(?!...)` | Negative lookahead. | `\d(?!px)` → digit not before `px` |
| `(?<=...)` | Positive lookbehind. | `(?<=\$)\d+` → number after `$` |
| `(?<!...)` | Negative lookbehind. | `(?<!\$)\d+` → number not after `$` |
