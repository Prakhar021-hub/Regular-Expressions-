```python
import re

print("="*20, "REGEX PRACTICE START", "="*20)

text_samples = [
    "Hello World!",
    "abc123",
    "Price is $100",
    "Email me at test.email@example.com",
    "My phone is +91-9876543210",
    "The cat sat on the mat.",
    "Visit https://www.example.com now!",
    "My birthday is 25-09-2025.",
    "Order ID: A1234-B567",
    "   lots   of   spaces   ",
    "UPPER and lower CASE",
    "Find me in educational context",
    "He runs quickly and beautifully.",
    "color or colour both are fine",
]

# Helper function
def show(pattern, text):
    print(f"Pattern: {pattern:25} | Text: {text:40} | Matches:", re.findall(pattern, text))

# -----------------------------
# 1. Basic matching
# -----------------------------
show(r"Hello", text_samples[0])
show(r"abc", text_samples[1])
show(r"\d+", text_samples[1])
show(r"\D+", text_samples[1])
show(r".", text_samples[0])

# -----------------------------
# 2. Anchors ^ and $
# -----------------------------
show(r"^Hello", text_samples[0])
show(r"World!$", text_samples[0])

# -----------------------------
# 3. Character classes
# -----------------------------
show(r"[aeiou]", text_samples[0])
show(r"[^aeiou]", text_samples[0])
show(r"[A-Z]", text_samples[10])
show(r"[a-z]", text_samples[10])
show(r"[0-9]", text_samples[1])

# -----------------------------
# 4. Quantifiers *, +, ?, {}
# -----------------------------
show(r"go*", "g go goo gooo")
show(r"go+", "g go goo gooo")
show(r"colou?r", text_samples[13])
show(r"a{2}", "caaabb")
show(r"a{2,4}", "caaaaabb")

# -----------------------------
# 5. Groups and alternation
# -----------------------------
show(r"(cat|dog)", text_samples[5])
show(r"(ab)+", "ababababx")
show(r"(?:ab)+", "ababababx")

# -----------------------------
# 6. Escaping special chars
# -----------------------------
show(r"\$", text_samples[2])
show(r"\.", text_samples[3])
show(r"\+", text_samples[4])

# -----------------------------
# 7. Word & whitespace
# -----------------------------
show(r"\w+", text_samples[1])
show(r"\W+", text_samples[1])
show(r"\s+", text_samples[9])
show(r"\S+", text_samples[9])

# -----------------------------
# 8. Boundaries
# -----------------------------
show(r"\bcat\b", text_samples[5])
show(r"\Bcat\B", text_samples[11])

# -----------------------------
# 9. Lookaheads & lookbehinds
# -----------------------------
show(r"\d+(?=px)", "10px 20px 30em")
show(r"\d+(?!px)", "10px 20px 30em")
show(r"(?<=\$)\d+", text_samples[2])
show(r"(?<!\$)\d+", "Code 123 not money")

# -----------------------------
# 10. Practical patterns
# -----------------------------
# Emails
show(r"[a-zA-Z0-9._%+-]+@[a-zA-Z0-9.-]+\.[a-z]{2,}", text_samples[3])

# URLs
show(r"https?://[^\s]+", text_samples[6])

# Phone numbers
show(r"\+91-\d{10}", text_samples[4])

# Dates (DD-MM-YYYY)
show(r"\d{2}-\d{2}-\d{4}", text_samples[7])

# Order ID
show(r"[A-Z]\d{4}-[A-Z]\d{3}", text_samples[8])

# HTML tags
show(r"<.*?>", "<html><body>Hello</body></html>")

# Hex color codes
show(r"#[0-9a-fA-F]{6}", "Color codes: #FF5733 #abcdef not#12345")

# Hashtags
show(r"#\w+", "Tweet with #hashtag and #Python3")

# Usernames (@name)
show(r"@\w+", "Follow me @openai and @chatgpt")

# IP addresses
show(r"\b\d{1,3}(\.\d{1,3}){3}\b", "Server at 192.168.1.1")

# Password strength (at least 1 upper, 1 lower, 1 digit, min 8 chars)
passwords = ["Test1234", "weak", "StrongPass1"]
for p in passwords:
    print(p, "->", bool(re.match(r"^(?=.*[a-z])(?=.*[A-Z])(?=.*\d).{8,}$", p)))

# Extract words ending with 'ly'
show(r"\w+ly\b", text_samples[12])

# Extract duplicate words
show(r"\b(\w+)\s+\1\b", "This is is a test test")

# Split text by multiple spaces
print("Split:", re.split(r"\s+", text_samples[9]))

# Replace digits with #
print("Replace digits:", re.sub(r"\d", "#", text_samples[1]))

# -----------------------------
# 11. Advanced usage
# -----------------------------
# Match camelCase words
show(r"[a-z]+[A-Z][a-z]+", "thisIsCamelCase anotherTest")

# Match snake_case words
show(r"\b[a-z]+(?:_[a-z]+)+\b", "snake_case test_example regex_code")

# Extract domain names
show(r"(?<=@)[\w.-]+", text_samples[3])

# Extract TLD
show(r"\.[a-z]{2,}", text_samples[3])

# Non-greedy vs greedy
show(r"<.*>", "<div>content</div>")
show(r"<.*?>", "<div>content</div>")

# Numbers with decimals
show(r"\d+(\.\d+)?", "Prices: 10, 20.5, 30")

# Scientific notation
show(r"[+-]?\d+(\.\d+)?[eE][+-]?\d+", "1.2e10 -3.5E-5")

# Palindrome-like sequence
show(r"(.)(.)\2\1", "abba noon deed")

# Extract capitalized words
show(r"\b[A-Z][a-z]+\b", "Alice went to London with Bob")

# Extract acronyms
show(r"\b[A-Z]{2,}\b", "NLP and AI are fields of CS")

print("="*20, "REGEX PRACTICE END", "="*20)
```
